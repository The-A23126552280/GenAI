```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="google/gemma-2-2b-it",
    token="hf_your_key_goes_here"
)
```

```python
question = "How would you make a samosa?"

# 1. Format the input using the standard chat template structure
messages = [
    {"role": "user", "content": question}
]

# 2. Apply the model's tokenizer chat template to get the correct formatting.
outputs = generator(
    messages,
    clean_up_tokenization_spaces=True,
    max_new_tokens=1024, # High ceiling so it never gets cut off mid-sentence
    do_sample=True,      # Enables creative generation strategies
    temperature=0.6      # Controls randomness (lower = more deterministic
)

# 3. Print the generated response
print(outputs[0]["generated_text"][-1]["content"])
```

```python
analyzer = pipeline(
    "text-classification",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)
```

```python
# 1. Define a list of diverse statements to test the model's performance
statements = [
    "The shipping was incredibly fast and the packaging kept everything safe!",
    "I wanted to love this, but it broke down within three days of normal use.",
    "It works fine, I guess. Nothing spectacular, but it does what it claims."
]

# 2. Pass the entire list to the analyzer for automated batch processing
outputs = analyzer(statements)

# 3. Loop through and display the evaluation outputs clearly
for text, result in zip(statements, outputs):
    print(f"Statement: \"{text}\"")
    print(f"Sentiment: {result['label']} (Confidence Score: {result['score']:.4f})\n")
```
